# COCO val2017 Sanity Check

这个 notebook 验证 COCO val2017 数据集（图像和关键点标注）是否下载并解析正确。

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pycocotools.coco import COCO

# 设置路径
DATA_ROOT = '../data/coco'
IMG_DIR = os.path.join(DATA_ROOT, 'val2017')
ANN_FILE = os.path.join(DATA_ROOT, 'annotations/person_keypoints_val2017.json')
```

```python
# 初始化 COCO api
coco = COCO(ANN_FILE)

# 获取所有包含 person 关键点的图像 ID
img_ids = coco.getImgIds(catIds=coco.getCatIds(catNms=['person']))
print(f"Total images with 'person' category in val2017: {len(img_ids)}")
```

```python
# 随机抽取 3 张图片可视化
num_samples = 3
sample_ids = random.sample(img_ids, num_samples)

fig, axes = plt.subplots(1, num_samples, figsize=(18, 6))

for i, img_id in enumerate(sample_ids):
    # 加载图像信息与图像文件
    img_info = coco.loadImgs(img_id)[0]
    img_path = os.path.join(IMG_DIR, img_info['file_name'])
    img = Image.open(img_path).convert('RGB')
    
    axes[i].imshow(img)
    axes[i].axis('off')
    axes[i].set_title(f"Image ID: {img_id}")
    
    # 获取这段图像上的人体姿态标注
    ann_ids = coco.getAnnIds(imgIds=img_id, catIds=coco.getCatIds(catNms=['person']), iscrowd=None)
    anns = coco.loadAnns(ann_ids)
    
    # 使用 coco api 在当前 subplot 上直接把标注的关键点画上去
    plt.sca(axes[i])
    coco.showAnns(anns)

plt.tight_layout()
plt.show()
```
