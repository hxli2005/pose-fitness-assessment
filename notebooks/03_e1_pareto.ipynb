# Pareto Frontier Visualization (E1 Experiment)
This notebook visualizes the trade-off between Model Accuracy (COCO AP) and Inference Speed (FPS), generating a standard bubble plot.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")

results_path = '../outputs/e1/results.csv'
if not os.path.exists(results_path):
    raise FileNotFoundError(f"Results file not found at {results_path}. Ensure you ran `scripts/eval_coco.py` first.")

df = pd.read_csv(results_path)
df
```

```python
plt.figure(figsize=(10, 7))

# Plotting bubbles where size reflects Params_M
# Create scalar mapping for bubble sizes
sizes = (df["Params_M"] + 1) * 20  # +1 to ensure base size, *20 for scaling visual impact

scatter = plt.scatter(x=df['FPS'], y=df['AP'], s=sizes, alpha=0.7, 
                      c=sns.color_palette("husl", len(df)), edgecolors='w', linewidth=1.5)

# Annotate points
for i, row in df.iterrows():
    plt.annotate(f"{row['model']} ({row['Params_M']:.1f}M)", 
                 (row['FPS'], row['AP']),
                 textcoords="offset points",
                 xytext=(0,10), 
                 ha='center',
                 fontsize=10,
                 fontweight='bold')

# Axis configurations
plt.title('Accuracy vs Speed Trade-off in 2D Pose Estimation', fontsize=15, pad=15)
plt.xlabel('Inference Speed (FPS → higher is better)', fontsize=12)
plt.ylabel('COCO Keypoint AP (Accuracy → higher is better)', fontsize=12)

# Optional: Log scale if x gaps are huge (e.g. 25fps vs 300fps)
# plt.xscale('log')

plt.margins(x=0.15, y=0.15)
plt.tight_layout()

# Save Plot
os.makedirs('../outputs/e1', exist_ok=True)
plt.savefig('../outputs/e1/pareto.png', dpi=300)
plt.show()
```
